```python
# import libraries. 
import os 
from langchain_cohere import CohereEmbeddings
from langchain_core.documents import Document
from langchain_community.vectorstores import FAISS
```

```python
# load the api keys 
os.environ['COHERE_API_KEY'] = "ADD YOUR OWN kE HERE"
```

```python
# create the document for the source. OR sample document. 
documents = [
    Document(page_content="LangChain makes it easy to work with LLMs."),
    Document(page_content="LangChain is used to build LLM based applications."),
    Document(page_content="Chroma is used to store and search document embeddings."),
    Document(page_content="Embeddings are vector representations of text."),
    Document(page_content="MMR helps you get diverse results when doing similarity search."),
    Document(page_content="LangChain supports Chroma, FAISS, Pinecone, and more."),
]
```

```python
# initialize the embeddig model. 
embedding_model = CohereEmbeddings(
    model = "embed-english-v3.0"
) 
```

```python
# create the vector store. 
vector_store = FAISS.from_documents(
    embedding = embedding_model, 
    documents = documents
)
```

```python
# create the retriever object. 
retriever = vector_store.as_retriever(
    search_type = 'mmr',
    search_kwargs = {
        'k' : 3, 
        'lambda_mult' : 1
    }
)
```

```python
# trigger the retiever. 
query = 'tell me about the Langchain ? '
docs = retriever.invoke(query)
```

```python
# print the results. 
print(docs)
```

```python
for page_num, doc in enumerate(docs): 
    print(f'-----Result {page_num + 1}-----\nContent : {doc.page_content}')
```
