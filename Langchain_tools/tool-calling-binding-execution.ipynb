```python
# Create tool. 

@tool
def multiply(a : int, b : int) -> int: 
    """ Here we take 2 numbers as input and this function will return the product of the two number"""
    return a * b
```

```python
print(multiply.invoke({'a' : 2, 'b' : 7}))
```

```python
# Tool Binding. 
```

```python
# create the llm object.
chat_model = ChatCohere(
    model="command-a-03-2025",
    temperature=0
)
```

```python
# calling the model. 
chat_model.invoke('Hi')
```

```python
llm_model_with_tools = chat_model.bind_tools([multiply])
```

```python
llm_model_with_tools.invoke('Can you multiply 2 with 7 ?')
```

```python
query = HumanMessage('Can you multiply 2 with 1000 ?')
```

```python
messages = [query]
```
