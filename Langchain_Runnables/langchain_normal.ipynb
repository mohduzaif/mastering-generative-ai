```python
# here we create the copy or trying to making the class to get better idea how chains internally depend on runnable class. 

import random

class NakliLLM: 

    def __init__(self):
        print('LLM created successfully')

    
    def predict(self, prompt): 

        response_list = [
            'AI is revolutionary in the upcoming years', 
            'IPL is a biggest league in the world', 
            'Delhi is the capital of India', 
            'World is under going through the crisis'
        ]
        return {'response' : random.choice(response_list)}
```

```python
llm = NakliLLM()
llm.predict('What is the capital of india?')
```

```python
# here we also create one more class for the same purpose. 
class NakliPromptTemplate: 

    def __init__(self, template, input_variables): 
        self.template = template
        self.input_variables = input_variables


    def format(self, input_dict): 
        return self.template.format(**input_dict)
```

```python
template = NakliPromptTemplate(
    template = 'Write the {length} poem about the {topic}', 
    input_variables = ['length', 'topic']
)
```

```python
prompt = template.format({
    'length' : 'short',
    'topic' : 'India'
})
print(prompt)
```

```python
llm = NakliLLM()
```

```python
llm.predict(prompt)
```

### `NOTE` : Now we trying to create the chains Intial phase of Langchain according to prospective of langchain developer they think we create alot of chain class, but after sometime they realise these are not flexible to communicate. 

```python
class NakliChain: 

    def __init__(self, prompt, llm): 
        self.prompt = prompt
        self.llm = llm 

    def run(self, input_dict): 
        final_prompt = self.prompt.format(input_dict)
        result = self.llm.predict(final_prompt)

        return result['response']
```

```python
llm = NakliLLM()
```

```python
template = NakliPromptTemplate(
    template = 'Write the {length} poem about the {topic}', 
    input_variables = ['length', 'topic']
)
```

```python
chain = NakliChain(template, llm)
```

```python
chain.run({
    'length' : 'short',
    'topic' : 'India' 
})
```

`NOTE : ` Here two step model calling is not possible, thats why they go for a runnable classes.
